```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
arquivo = 'Tabela_13_Meio_de_transporte.xlsx'

df = pd.read_excel(
    arquivo,
    sheet_name='Exemplo gráfico Brasil',
    header=1
)

df
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

cores = [
    '#ff9999',
    '#66b3ff',
    '#99ff99',
    '#ffcc99',
    '#c2c2f0',
    '#ffb3e6'
]

grupos = ['Branca', 'Preta ou parda', 'Indígena']

for i, col in enumerate(grupos):
    axes[i].pie(
        df[col],
        labels=df['Meio de transporte/cor ou raça'],
        autopct='%1.3f%%',
        startangle=180,
        colors=cores,
        wedgeprops={
            'edgecolor': 'white',
            'linewidth': 1.2
        }
    )

    axes[i].set_title(
        f'Cor/Raça: {col}',
        fontsize=12,
        fontweight='bold'
    )

plt.suptitle(
    'Meios de transporte utilizados por mulheres no Brasil, segundo cor ou raça - 2022',
    fontsize=14,
    fontweight='bold'
)

plt.tight_layout()

plt.savefig(
    'grafico_transporte.png',
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

```python
df_br = pd.read_excel(
    arquivo,
    sheet_name='BR GR UF MU',
    header=None
)

df_br.head()
```

```python
colunas = [
    'Localidade',
    
    'Branca - A pé',
    'Branca - Bicicleta',
    'Branca - Motocicleta ou Mototáxi',
    'Branca - Automóvel, táxi ou assemelhados',
    'Branca - Transporte coletivo',
    'Branca - Outros',
    
    'Preta ou Parda - A pé',
    'Preta ou Parda - Bicicleta',
    'Preta ou Parda - Motocicleta ou Mototáxi',
    'Preta ou Parda - Automóvel, táxi ou assemelhados',
    'Preta ou Parda - Transporte coletivo',
    'Preta ou Parda - Outros',
    
    'Indígena - A pé',
    'Indígena - Bicicleta',
    'Indígena - Motocicleta ou Mototáxi',
    'Indígena - Automóvel, táxi ou assemelhados',
    'Indígena - Transporte coletivo',
    'Indígena - Outros'
]

dados = df_br.iloc[3:].copy()
dados.columns = colunas

for coluna in colunas[1:]:
    dados[coluna] = pd.to_numeric(
        dados[coluna],
        errors='coerce'
    )

dados.head()
```

```python
estados = [
    'Rondônia',
    'Acre',
    'Amazonas',
    'Roraima',
    'Pará',
    'Amapá',
    'Tocantins',
    'Maranhão',
    'Piauí',
    'Ceará',
    'Rio Grande do Norte',
    'Paraíba',
    'Pernambuco',
    'Alagoas',
    'Sergipe',
    'Bahia',
    'Minas Gerais',
    'Espírito Santo',
    'Rio de Janeiro',
    'São Paulo',
    'Paraná',
    'Santa Catarina',
    'Rio Grande do Sul',
    'Mato Grosso do Sul',
    'Mato Grosso',
    'Goiás',
    'Distrito Federal'
]

df_estados = dados[
    dados['Localidade'].isin(estados)
].copy()

df_estados[['Localidade']]
```

```python
df_melt = dados.melt(
    id_vars='Localidade',
    var_name='Cor_Raca_Meio',
    value_name='Percentual'
)

df_melt.head()
```

```python
df_melt[['Cor/Raça', 'Meio de Transporte']] = (
    df_melt['Cor_Raca_Meio']
    .str.split(' - ', n=1, expand=True)
)

df_melt.head()
```

```python
coletivo_estados = df_melt[
    (df_melt['Localidade'].isin(estados)) &
    (df_melt['Meio de Transporte'] == 'Transporte coletivo')
]

resultado_coletivo = (
    coletivo_estados
    .groupby('Localidade')['Percentual']
    .mean()
    .sort_values(ascending=False)
)

resultado_coletivo
```

O estado que apresenta o maior percentual médio de mulheres utilizando **transporte coletivo** é o **Rio de Janeiro**, com aproximadamente **52,23%**.

Esse valor corresponde à média dos percentuais dos três grupos de cor ou raça presentes na tabela: Branca, Preta ou Parda e Indígena.

```python
resultado_coletivo.sort_values().head(1)
```

O estado que apresenta o menor percentual médio de mulheres utilizando **transporte coletivo** é **Rondônia**, com aproximadamente **4,33%**.

O valor representa a média dos percentuais dos três grupos de cor ou raça analisados.

```python
df_municipios = dados[
    dados['Localidade'].str.contains(
        r'\([A-Z]{2}\)$',
        regex=True,
        na=False
    )
].copy()

df_municipios[['Localidade']].head()
```

```python
bicicleta = df_municipios.melt(
    id_vars='Localidade',
    var_name='Cor_Raca_Meio',
    value_name='Percentual'
)

bicicleta[['Cor/Raça', 'Meio de Transporte']] = (
    bicicleta['Cor_Raca_Meio']
    .str.split(' - ', n=1, expand=True)
)

bicicleta = bicicleta[
    bicicleta['Meio de Transporte'] == 'Bicicleta'
]

resultado_bicicleta = (
    bicicleta
    .groupby('Localidade')['Percentual']
    .mean()
    .sort_values(ascending=False)
)

resultado_bicicleta.head(10)
```

A cidade que apresenta o maior percentual médio de mulheres utilizando **bicicleta** é **Tuiuti (SP)**, com **100%**.

Nesse caso, o percentual médio dos três grupos de cor ou raça chega a 100%.

```python
carro = df_municipios.melt(
    id_vars='Localidade',
    var_name='Cor_Raca_Meio',
    value_name='Percentual'
)

carro[['Cor/Raça', 'Meio de Transporte']] = (
    carro['Cor_Raca_Meio']
    .str.split(' - ', n=1, expand=True)
)

carro = carro[
    carro['Meio de Transporte'] == 'Automóvel, táxi ou assemelhados'
]

resultado_carro = (
    carro
    .groupby('Localidade')['Percentual']
    .mean()
    .sort_values(ascending=False)
)

resultado_carro.head(10)
```

A cidade que apresenta o maior percentual médio de mulheres utilizando **carro, táxi ou assemelhados** é **Bom Jesus do Oeste (SC)**, com aproximadamente **86,05%**.

O resultado considera a média dos percentuais dos três grupos de cor ou raça disponíveis na tabela.
